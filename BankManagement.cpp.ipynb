```python
%%writefile BankManagementSystem.cpp

#include <iostream>
#include <fstream>
#include <string>
using namespace std;

class BankAccount {
public:
    int accountNo;
    string name;
    double balance;

    void createAccount() {
        cout << "\nEnter Account Number: ";
        cin >> accountNo;
        cin.ignore();

        cout << "Enter Customer Name: ";
        getline(cin, name);

        cout << "Enter Initial Deposit: ";
        cin >> balance;

        if (balance < 0) {
            balance = 0;
            cout << "Invalid amount. Balance set to 0.\n";
        }

        saveAccount();

        cout << "\nAccount created successfully!\n";
    }

    void saveAccount() {
        ofstream file("bank_records.txt", ios::app);

        file << accountNo << "|"
             << name << "|"
             << balance << endl;

        file.close();
    }

    bool findAccount(int accNo) {
        ifstream file("bank_records.txt");

        string line;

        while (getline(file, line)) {
            size_t p1 = line.find('|');
            size_t p2 = line.find('|', p1 + 1);

            if (p1 == string::npos || p2 == string::npos)
                continue;

            int acc = stoi(line.substr(0, p1));

            if (acc == accNo) {
                accountNo = acc;
                name = line.substr(p1 + 1, p2 - p1 - 1);
                balance = stod(line.substr(p2 + 1));

                file.close();
                return true;
            }
        }

        file.close();
        return false;
    }

    void updateAccount() {
        ifstream file("bank_records.txt");
        ofstream temp("temp.txt");

        string line;

        while (getline(file, line)) {
            size_t p1 = line.find('|');
            size_t p2 = line.find('|', p1 + 1);

            if (p1 == string::npos || p2 == string::npos)
                continue;

            int acc = stoi(line.substr(0, p1));

            if (acc == accountNo) {
                temp << accountNo << "|"
                     << name << "|"
                     << balance << endl;
            } else {
                temp << line << endl;
            }
        }

        file.close();
        temp.close();

        remove("bank_records.txt");
        rename("temp.txt", "bank_records.txt");
    }

    void deposit() {
        double amount;

        cout << "\nEnter Deposit Amount: ";
        cin >> amount;

        if (amount <= 0) {
            cout << "Invalid deposit amount.\n";
            return;
        }

        balance += amount;
        updateAccount();

        cout << "Amount deposited successfully.\n";
        cout << "New Balance: Rs. " << balance << endl;
    }

    void withdraw() {
        double amount;

        cout << "\nEnter Withdrawal Amount: ";
        cin >> amount;

        if (amount <= 0) {
            cout << "Invalid withdrawal amount.\n";
            return;
        }

        if (amount > balance) {
            cout << "Insufficient balance.\n";
            return;
        }

        balance -= amount;
        updateAccount();

        cout << "Amount withdrawn successfully.\n";
        cout << "Remaining Balance: Rs. " << balance << endl;
    }

    void checkBalance() {
        cout << "\nAccount Number: " << accountNo << endl;
        cout << "Customer Name: " << name << endl;
        cout << "Current Balance: Rs. " << balance << endl;
    }
};

int main() {
    int choice;
    int accNo;

    while (true) {
        cout << "\n========== BANK MANAGEMENT SYSTEM ==========\n";
        cout << "1. Create Account\n";
        cout << "2. Deposit Money\n";
        cout << "3. Withdraw Money\n";
        cout << "4. Check Balance\n";
        cout << "5. Exit\n";
        cout << "============================================\n";

        cout << "Enter your choice: ";
        cin >> choice;

        BankAccount account;

        switch (choice) {

        case 1:
            account.createAccount();
            break;

        case 2:
            cout << "\nEnter Account Number: ";
            cin >> accNo;

            if (account.findAccount(accNo))
                account.deposit();
            else
                cout << "Account not found.\n";

            break;

        case 3:
            cout << "\nEnter Account Number: ";
            cin >> accNo;

            if (account.findAccount(accNo))
                account.withdraw();
            else
                cout << "Account not found.\n";

            break;

        case 4:
            cout << "\nEnter Account Number: ";
            cin >> accNo;

            if (account.findAccount(accNo))
                account.checkBalance();
            else
                cout << "Account not found.\n";

            break;

        case 5:
            cout << "\nThank you for using Bank Management System!\n";
            return 0;

        default:
            cout << "Invalid choice. Please try again.\n";
        }
    }

    return 0;
}
```
